## **Nulls** [8 marks]

[1+1+3+2+1]

Find out how many nulls in each column

```python
# to check null values in each column
sales.isnull().sum()
```

```python
sales.isnull().sum().sort_values(ascending=False)
```

Find all those rows where brand is null

```python
sales[sales['brand'].isnull()]
```

Fill all the nulls in brand by extracting the brand name from product_name

(hint: look into str.rsplit)

```python
# Fill null brand values by extracting brand from product_name
sales['brand'] = sales['brand'].fillna(
    sales['product_name'].str.rsplit(' ', n=1).str[0]
)
```

```python
sales['brand'].isnull().sum()
```

Fill all the nulls in payment_method with the most frequently occuring payment method of the WHOLE column

```python
# find the most frequent payment method
most_frequent_method = sales['payment_method'].mode()[0]
print("Most frequent payment method:", most_frequent_method)
```

```python
# Fill missing values with that method
sales['payment_method'] = sales['payment_method'].fillna(most_frequent_method)
```

Check total nulls specifically in the payment_method column to see if all were successfully removed

```python
sales['payment_method'].isnull().sum()
```

## **Messy Data** [5 marks]

[1 + 3 + 1]

Find all the unique subcategories

```python
sales['&subcategory'].unique()
```

Clean up the messy text in the subcategory column

(hint: look for a .str function for this)

```python
# Clean up the messy text in the subcategory column
sales['&subcategory'] = (
    sales['&subcategory']
    .str.replace('&', '', regex=False)       # remove '&'
    .str.replace('[0-9]', '', regex=True)    # remove digits
    .str.strip()                             # remove leading/trailing spaces
    .str.title()                             # capitalize first letter
)
```

```python
sales['&subcategory'] = sales['&subcategory'].astype('string')
```

```python
# Create a dictionary of wrong → correct spellings
corrections = {
    'Shos': 'Shoes',
    'Bootis': 'Booties',
    'Swatrs': 'Sweaters',
    'Romprs': 'Rompers',
    'Drsss': 'Dresses',
    'Jwlry': 'Jewelry',
    'Snakrs': 'Sneakers',
    'Watchs': 'Watches',
    'Swatrs': 'Sweaters',
    'Tops': 'Tops',
}

sales['&subcategory'] = sales['&subcategory'].replace(corrections)
```

```python
sales['&subcategory'].unique()
```

Rename the subcategory column to remove messy text from there as well

```python
sales.rename(columns={'&subcategory': 'subcategory'}, inplace=True)
```

```python
sales['subcategory'].unique()
```

### **Total Sales** [3 marks]

Create a total_sale column (remember to factor in discount). It should be rounded to 2 decimal places

```python
sales['total_sale'] = (
    sales['price'] * sales['quantity'] * (1 - sales['discount_percent'] / 100)
).round(2)
```

```python
sales[['price', 'quantity', 'discount_percent', 'total_sale']].head()
```

### **Groupby and Transform** [6 marks]

[3 + 3]

For each null in the price column, replace it with the average price of the subcategory occuring at that row.

(hint: use transform to get a column of means that is mappable to our dataframe and use that with fillna)


```python
sales['price'] = sales['price'].fillna(
    sales.groupby('subcategory')['price'].transform('mean')
)
```

```python
sales['price'].isnull().sum()
```

```python
sales['price'].describe()
```

Find out how many unique brands in each subcategory.

Store this result in a variable called brands_per_subcategory

```python
brands_per_subcategory = sales.groupby('subcategory')['brand'].nunique()
brands_per_subcategory
```

## **Plotting** [7 marks]

[2+1+2+2]

Plot a count plot of brands.

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(12,6))  # set figure size
sns.countplot(data=sales, x='brand', order=sales['brand'].value_counts().index)

plt.title("Count of Products by Brand", fontsize=14)
plt.xlabel("Brand")
plt.ylabel("Count")
plt.xticks(rotation=90)  # rotate x-labels for readability
plt.show()
```

In the above plot, try and rearrange the bars in **ascending** order by height (i.e. by count)

```python
plt.figure(figsize=(12,6))

sns.countplot(
    data=sales,
    x='brand',
    order=sales['brand'].value_counts(ascending=True).index  # ascending order
)

plt.title("Count of Products by Brand (Ascending Order)", fontsize=14)
plt.xlabel("Brand")
plt.ylabel("Count")
plt.xticks(rotation=90)
plt.show()
```
